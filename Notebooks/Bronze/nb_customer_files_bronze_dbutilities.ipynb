```python
dbutils.widgets.text(
    name="catalog_name",
    defaultValue="intellibi_catalog",
    label="catalog_name"
)

dbutils.widgets.text(
    name="schema_name",
    defaultValue="intellibi_bronze",
    label="schema_name"
)

dbutils.widgets.text(
    name="file_location",
    defaultValue="/Volumes/intellibi_catalog/intellibi_schema/intellibi_data_engineering_volume/src_files/daily_customers/",
    label="file_location"
)
```

```python
catalog = dbutils.widgets.get("catalog_name")
schema = dbutils.widgets.get("schema_name")
filelocation= dbutils.widgets.get("file_location")
table_name=f"{catalog}.{schema}.rawCustomer"


```

```python
print(table_name)
print(filelocation)
```

```python
sql_query= f"""copy into {table_name}
from (
select 
customer_id,
customer_name,
email ,
 phone ,
address ,
city ,
state ,
country ,
zip_code ,
segment ,
current_timestamp() as ingest_ts 
from '{filelocation}' )
FILEFORMAT = CSV
FORMAT_OPTIONS ('header'='true')
COPY_OPTIONS ('mergeSchema'='false')"""
df=spark.sql(sql_query)
df.show()
```
