```python
%sql
--  scd type- 1 --> it is upsert (insert or update  based on the condition) (implement by MERGE Statement)


MERGE INTO intellibi_catalog.intellibi_gold.dfMonthlySales AS target
USING (
  SELECT *
  FROM intellibi_catalog.intellibi_silver.dfMonthlySales
  QUALIFY ROW_NUMBER() OVER (PARTITION BY sale_id ORDER BY sale_id) = 1
) AS source
ON source.sale_id = target.sale_id
WHEN MATCHED THEN UPDATE SET *
WHEN NOT MATCHED THEN INSERT *;

-- --resolve issue :- 
-- ISSUE : id is same and data is same than also it is updating , showing num_affected_rows, ideally it should not be there. How to handle it ?
-- Solution : it can be handled using sql and using pyspark too. There are many ways to do this. 
```
