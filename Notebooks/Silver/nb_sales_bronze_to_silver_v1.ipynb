The following operations are performaed in silver tables :-
- Schema enforcement / column schema evolution 
- Handling of null and missing values- 
-  Data deduplication- 
-  Resolution of out-of-order and late-arriving data issues- 
-  Data quality checks and enforcement- 
-  Schema evolution- 
-  Type casting / data type handling
- case conversion 
- Trimming 
- maintain order of the data 
- JOINS




Start modeling data:- 

It is common to start performing data modeling in the silver layer, including choosing how to represent heavily nested or semi-structured data:
- Use VARIANT data type.
- Use JSON strings.
- Create structs, maps, and arrays.
- Flatten schema or normalize data into multiple tables.

```python
%sql
insert into intellibi_catalog.intellibi_silver.cleanedMonthlySales (
select distinct
cast(trim(sale_id) as int) as sale_id,
upper(trim(product) ),
upper(trim(category) ),
cast(trim(quantity) as int) as quantity,
cast(trim(price) as double) as price,
to_date(trim(sale_date) , 'yyyy-MM-dd') as sale_date, 
upper(trim(region)) ,
current_timestamp() as ingest_ts
from intellibi_catalog.intellibi_bronze.rawmonthlysales
where intellibi_catalog.intellibi_bronze.rawmonthlysales.ingest_ts > 
                    (select coalesce( max(ingest_ts), to_date('1990-01-01','yyyy-MM-dd') ) from intellibi_catalog.intellibi_silver.cleanedMonthlySales)
order by cast(trim(sale_id) as int)
)
```
