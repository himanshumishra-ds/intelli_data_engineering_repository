```python
%python
df_monthly_sales=spark.sql("select * from intellibi_catalog.intellibi_bronze.dfMonthlySales")
```

```python
# trim all the columns values and make all the values in upper case 
df_monthly_sales_clean = df_monthly_sales.selectExpr(
    "UPPER(TRIM(sale_id)) AS sale_id",
    "UPPER(TRIM(product)) AS product",
    "UPPER(TRIM(category)) AS category",
    "UPPER(TRIM(quantity)) AS quantity",
    "UPPER(TRIM(price)) AS price",
    "UPPER(TRIM(sale_date)) AS sale_date",
    "UPPER(TRIM(region)) AS region"
)



df_monthly_sales_clean.display()
```
